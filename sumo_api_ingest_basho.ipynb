```python
import json
import time

import requests

# GET /api/basho/:bashoId/torikumi/:division/:day
base_url = "https://sumo-api.com/api/basho"

first_year = 2025
last_year = 2025
years = range(first_year, last_year + 1)
first_month = 1
last_month = 11
months = ["{:02d}".format(month) for month in range(first_month, last_month + 1, 2)]
first_day = 1
last_day = 15
days = ["{:02d}".format(day) for day in range(first_day, last_day + 1)]
divisions = ["makuuchi", "juryo", "makushita", "sandanme", "jonidan", "jonokuchi"]


# tournaments in range where we either already have data, or where there is no data
excluded_basho_ids = {
    # already have data
    "202501",
    "202503",
    # there is no data
    "202505",
    "202507",
    "202509",
    "202511",
}

params = [
    (str(year) + month, division, day)
    for year in years
    for month in months
    for day in days
    for division in divisions
]
params = [param for param in params if param[0] not in excluded_basho_ids]
```

```python
failed_urls: list[str] = []
failed_urls_path = "failed_urls_basho.json"
for param in params:
    basho_id, division, day = param
    url = f"{base_url}/{basho_id}/torikumi/{division}/{day}"

    try:
        response = requests.get(url)
        if response.status_code == 200:
            filename = f"raw_data/basho/{basho_id}_{division}_{day}.json"
            with open(filename, "w") as f:
                json.dump(response.json(), f, indent=2)
        else:
            print(f"Error fetching {url}, status code: {response.status_code}")
            failed_urls += url

    except requests.exceptions.RequestException as e:
        print(f"Error fetching {url}, error: {e}")
        failed_urls += url

    time.sleep(1)

with open(failed_urls_path, "w") as f:
    json.dump(failed_urls, f)

print(failed_urls)
print("DONE!")
```
